# NeuralOps Attention Surgery + LoRA

A minimal end-to-end DistilBERT workflow: replace attention, inject LoRA, run a KD-assisted update, and perform inference.
### 1.1 Low-Rank Adaptation (LoRA) Mathematical Contract
Parameter-Efficient Fine-Tuning (PEFT) freezes the underlying foundation model parameters $\mathbf{W}_0 \in \mathbb{R}^{d_{\mathrm{in}} \times d_{\mathrm{out}}}$ and injects rank decomposition matrices $\mathbf{B} \in \mathbb{R}^{d_{\mathrm{in}} \times r}$ and $\mathbf{A} \in \mathbb{R}^{r \times d_{\mathrm{out}}}$ with $r \ll \min(d_{\mathrm{in}}, d_{\mathrm{out}})$:

$$\mathbf{h} = \mathbf{W}_0 \mathbf{x} + \Delta \mathbf{W} \mathbf{x} = \mathbf{W}_0 \mathbf{x} + \frac{\alpha}{r} (\mathbf{B} \mathbf{A}) \mathbf{x}
$$

where $\alpha$ is a scaling hyperparameter that stabilizes the initialization when changing the intrinsic rank $r$.

### 1.2 Comparison of Subquadratic Kernel Formulations
| Attention Mechanism | Time Complexity | KV-State Dimension | Gated Non-Linearity | Hardware Efficiency |
| :--- | :---: | :---: | :---: | :---: |
| **Softmax Attention (Standard)** | $\mathcal{O}(S^2)$ | $\mathcal{O}(S)$ (Unbounded) | Softmax Row-Norm | Memory-Bound (HBM) |
| **Linear Attention (Katharopoulos)** | $\mathcal{O}(S)$ | $\mathcal{O}(D^2)$ (Fixed) | ELU + 1 | Compute-Bound (GEMM) |
| **Gated Linear Attention (GLA)** | $\mathcal{O}(S)$ | $\mathcal{O}(D^2)$ (Fixed) | Sigmoid Data-Dependent Gate | Chunkwise Parallel |
| **Mamba / Selective SSM** | $\mathcal{O}(S)$ | $\mathcal{O}(D \times N)$ (State) | HiPPO Continuous-time Discretization | Hardware Prefix Scan |

```text
+---------------------------------------------------------------------------------------------------+
|                               LoRA Surgical Bridge Topology                                       |
|                                                                                                   |
|  Input Hidden Vector x ---> [Frozen Pre-trained Weight W_0] ----------------------------(+)-----> Output
|                       │                                                                  ▲        |
|                       └───> [Down-Projection B: d_in -> r] ---> [Up-Proj A: r -> d_out] ─┘        |
|                                      (Trainable Adapter Parameters)                               |
+---------------------------------------------------------------------------------------------------+
```


````python
import copy
import torch
import torch.nn.functional as F
import neuralops
from transformers import DistilBertConfig, DistilBertForSequenceClassification

torch.manual_seed(42)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
config = DistilBertConfig(
    vocab_size=101, dim=64, hidden_dim=128, n_layers=1, n_heads=4, num_labels=2
)
teacher = DistilBertForSequenceClassification(config).to(device).eval()
student = copy.deepcopy(teacher)
student = neuralops.replace_attention(
    student, "cosformer", copy_weights=False,
    inject_lora=True, lora_rank=4, lora_alpha=8.0, lora_dropout=0.0,
).to(device)
print("device:", device, "attention:", student.config.neuralops_attention)
````

````python
attention_names = [
    name for name, module in student.named_modules()
    if "attention" in name and module.__class__.__name__ == "UniversalAttentionWrapper"
]
lora_names = [name for name, _ in student.named_parameters() if "lora_" in name.lower()]
assert attention_names and lora_names
print("replaced attention:", attention_names)
print("LoRA parameters:", len(lora_names))
````

````python
input_ids = torch.randint(0, config.vocab_size, (2, 8), device=device)
attention_mask = torch.ones_like(input_ids)
labels = torch.randint(0, 2, (2,), device=device)
with torch.no_grad():
    teacher_logits = teacher(input_ids=input_ids, attention_mask=attention_mask).logits
student.train()
optimizer = torch.optim.AdamW([p for p in student.parameters() if p.requires_grad], lr=1e-3)
student_logits = student(input_ids=input_ids, attention_mask=attention_mask).logits
from neuralops.trainer.distillation.multi_level_kd import LogitDistillationLoss
kd_loss = LogitDistillationLoss(mode="forward_kl", temperature=2.0)(
    student_logits, teacher_logits.detach()
)
loss = kd_loss + F.cross_entropy(student_logits, labels)
optimizer.zero_grad(set_to_none=True)
loss.backward()
optimizer.step()
grads = [p.grad for p in student.parameters() if p.requires_grad and p.grad is not None]
assert grads and all(torch.isfinite(g).all() for g in grads)
print("KD loss:", float(kd_loss.detach()), "total loss:", float(loss.detach()), "finite gradients:", True)
````

````python
student.eval()
with torch.no_grad():
    prediction = student(input_ids=input_ids, attention_mask=attention_mask).logits.argmax(dim=-1)
assert torch.isfinite(prediction.float()).all()
print("predictions:", prediction.tolist())
````

## Defensive Engineering & Common MLSys Anti-Patterns

When implementing high-performance attention surgery and compression pipelines, watch out for:
- **Anti-Pattern 1: Un-synchronized CUDA Kernels during Timings**: Using `time.time()` without `torch.cuda.synchronize()` measures asynchronous CPU kernel launch latency rather than GPU execution.
- **Anti-Pattern 2: FP16 Intermediate State Divergence**: Accumulating linear recurrence states in FP16 produces exponential loss spikes due to gradient underflow.
- **Anti-Pattern 3: In-Place Weight Mutation without Healing**: Performing aggressive SVD truncation or 4-bit quantization without downstream calibration degrades perplexity beyond recovery.

## In-Place Module Transplant Surgery & Frobenius Projection Residue

When replacing standard multi-head self-attention with linear attention (e.g., Cosformer, GLA) in large-scale transformer backbones, the projection matrices $\mathbf{W}_q, \mathbf{W}_k, \mathbf{W}_v, \mathbf{W}_o$ must be surgically mapped. We quantify initial subspace distortion via the relative Frobenius norm residue:

85970\mathcal{R}_F(\mathbf{W}_{\mathrm{target}}, \mathbf{W}_{\mathrm{source}}) = \frac{\mathrm{\|} \mathbf{W}_{\mathrm{source}} - \mathbf{W}_{\mathrm{target}} \mathrm{\|}_F}{\mathrm{\|} \mathbf{W}_{\mathrm{source}} \mathrm{\|}_F + \epsilon}85970

NeuralOps ensures exact subspace preservation at surgery time (=0$) through structural weight alignment:

85970\mathbf{W}_{\mathrm{target}} \leftarrow \mathbf{W}_{\mathrm{source}} \implies \mathcal{R}_F = 085970

Below, we perform an automated in-place module transplant and compute the exact Frobenius residue across all attention projections.

````python
import torch
import torch.nn as nn
from neuralops.core.attention_surgery import compute_reconstruction_error

# Simulate legacy attention projections
d_model = 128
torch.manual_seed(42)
source_q = nn.Linear(d_model, d_model, bias=False)
target_q = nn.Linear(d_model, d_model, bias=False)

# Compute initial random error
err_pre = compute_reconstruction_error(source_q.weight, target_q)

# Execute in-place transplant surgery
with torch.no_grad():
    target_q.weight.copy_(source_q.weight)

# Verify zero-tolerance post-transplant error
err_post = compute_reconstruction_error(source_q.weight, target_q)

print("=" * 65)
print("[Module Transplant Surgery] In-Place Subspace Verification:")
print(f"  Pre-Surgery Relative Frobenius Residue:  {err_pre:.4f}")
print(f"  Post-Surgery Relative Frobenius Residue: {err_post:.4e}")
assert err_post < 1e-6, "Transplant surgery failed zero-tolerance parity!"
print("  In-place attention projection transplant passed with zero error.")
print("=" * 65)
````

## Rank-$ LoRA Parameterization & Directional Energy Ratio

Low-Rank Adaptation restricts parameter updates to a low-dimensional manifold:

85970\Delta \mathbf{W} = \frac{\alpha}{r} \mathbf{B} \mathbf{A}, \quad \mathbf{B} \in \mathbb{R}^{d_{\mathrm{out}} \times r}, \; \mathbf{A} \in \mathbb{R}^{r \times d_{\mathrm{in}}}85970

To ensure that surgery does not corrupt pre-trained representations at initialization ( = 0$):
- $\mathbf{A} \sim \mathcal{N}(0, \sigma^2)$ (random Gaussian exploration)
- $\mathbf{B} = \mathbf{0}$ (strict zero initialization)

This guarantees that $\Delta \mathbf{W} = \mathbf{0}$ and $\mathbf{W}_{\mathrm{eff}} = \mathbf{W}_{\mathrm{base}}$, preserving initial functional behavior:

85970\mathrm{\|} \mathbf{h}_{\mathrm{LoRA}} - \mathbf{h}_{\mathrm{base}} \mathrm{\|}_F = 085970

The cell below audits the rank spectrum and singular energy distribution of $\mathbf{B} \mathbf{A}$ across training steps.

````python
import torch
import torch.nn as nn

class SurgicalLoRAInjection(nn.Module):
    def __init__(self, linear: nn.Linear, rank: int = 8, alpha: float = 16.0):
        super().__init__()
        self.linear = linear
        self.linear.weight.requires_grad = False
        self.rank = rank
        self.scaling = alpha / rank
        
        self.lora_A = nn.Parameter(torch.randn(rank, linear.in_features) * (1.0 / rank))
        self.lora_B = nn.Parameter(torch.zeros(linear.out_features, rank))
        
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear(x) + (x @ self.lora_A.t()) @ self.lora_B.t() * self.scaling

base_mod = nn.Linear(128, 128, bias=False)
surg_lora = SurgicalLoRAInjection(base_mod, rank=8, alpha=16.0)

x_probe = torch.randn(4, 32, 128)
out_base = base_mod(x_probe)
out_surg = surg_lora(x_probe)

delta_init = torch.norm(out_surg - out_base, p="fro").item()
print("=" * 65)
print("[Rank-k LoRA Verification] Initialization & Energy Spectrum:")
print(f"  Initial Output Frobenius Delta: {delta_init:.6f}")
assert delta_init == 0.0, "Surgical LoRA injection must preserve identity at t=0!"

# Simulate single micro-step update to inspect singular value spectrum of Delta W
surg_lora.lora_B.data.normal_(0.0, 0.01)
delta_W = (surg_lora.lora_B @ surg_lora.lora_A) * surg_lora.scaling
U, S_vals, V = torch.linalg.svd(delta_W)
active_rank = (S_vals > 1e-5).sum().item()
print(f"  Configured Rank: {surg_lora.rank} | Measured Singular Rank: {active_rank}")
assert active_rank <= surg_lora.rank, "Active rank exceeds structural rank budget!"
print("  LoRA rank-8 parameterization confirmed strictly rank-bounded.")
print("=" * 65)
````

````python
import torch
from neuralops.ops.triton.lightning_scan import lightning_scan, lightning_scan_pytorch

B, H, T, D = 2, 4, 64, 32
torch.manual_seed(42)
q = torch.randn(B, H, T, D)
k = torch.randn(B, H, T, D)
v = torch.randn(B, H, T, D)

# Execute Level 3 Reference Fallback
out_ref = lightning_scan_pytorch(q, k, v, s=0.1)

# Execute Multi-Tier Dispatcher
out_disp = lightning_scan(q, k, v, s=0.1)

max_diff = torch.max(torch.abs(out_disp - out_ref)).item()
print("=" * 65)
print("[Multi-Tier Dispatch Verification]:")
print(f"  Dispatched Output Shape: {list(out_disp.shape)}")
print(f"  Max Difference vs Golden Reference: {max_diff:.6e}")
assert max_diff < 1e-4, "Dispatch output deviates from reference!"
print("  Multi-tier dispatch parity gate PASSED (atol < 1e-4).")
print("=" * 65)
````
